## Cell 2: Import the packages

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

print("NumPy:", np.__version__)
print("pandas:", pd.__version__)
print("scikit-learn:", sklearn.__version__)
print("Environment setup successful!")
```

## Cell 3: Run a small experiment

```python
models = ["Classical ML", "XLM-R", "LLM"]
macro_f1 = [0.460, 0.664, 0.580]

results = pd.DataFrame({
    "Model": models,
    "Macro-F1": macro_f1
})

results
```

## Cell 4: Create a graph

```python
plt.figure(figsize=(8, 4))

plt.bar(
    results["Model"],
    results["Macro-F1"],
    color=["#4C78A8", "#2A9D8F", "#F4A261"]
)

plt.ylim(0, 1)
plt.ylabel("Macro-F1")
plt.title("Example Model Comparison")
plt.show()
```
