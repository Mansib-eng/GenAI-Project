## 1. Create vectors and inspect their shapes

A vector is an ordered collection of numbers. In machine learning, a vector can represent a text, document, image, or data record.

````python
vector_a = np.array([1.0, 2.0, 3.0])
vector_b = np.array([4.0, 5.0, 6.0])
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6],
])

print('Vector A:', vector_a)
print('Vector A shape:', vector_a.shape)
print('Vector A dimensions:', vector_a.ndim)
print('Matrix:\n', matrix)
print('Matrix shape:', matrix.shape)
print('Matrix dimensions:', matrix.ndim)

assert vector_a.shape == (3,)
assert matrix.shape == (2, 3)
````

## 2. Dot product

For vectors $A=[a_1,a_2,\ldots,a_n]$ and $B=[b_1,b_2,\ldots,b_n]$:

$$A \cdot B = a_1b_1 + a_2b_2 + \cdots + a_nb_n$$

For $A=[1,2,3]$ and $B=[4,5,6]$:

$$A \cdot B=(1\times4)+(2\times5)+(3\times6)=32$$

````python
# Manual calculation with ordinary Python
manual_products = [a * b for a, b in zip(vector_a, vector_b)]
manual_dot_product = sum(manual_products)

# NumPy calculation
numpy_dot_product = np.dot(vector_a, vector_b)

print('Element-wise products:', manual_products)
print('Manual dot product:', manual_dot_product)
print('NumPy dot product:', numpy_dot_product)

assert manual_dot_product == 32
assert np.isclose(manual_dot_product, numpy_dot_product)
````

## 3. Vector magnitude

The magnitude, or Euclidean norm, of vector $A$ is:

$$\lVert A \rVert=\sqrt{a_1^2+a_2^2+\cdots+a_n^2}$$

````python
manual_magnitude_a = math.sqrt(sum(value ** 2 for value in vector_a))
manual_magnitude_b = math.sqrt(sum(value ** 2 for value in vector_b))

numpy_magnitude_a = np.linalg.norm(vector_a)
numpy_magnitude_b = np.linalg.norm(vector_b)

print('Manual magnitude of A:', manual_magnitude_a)
print('NumPy magnitude of A:', numpy_magnitude_a)
print('Manual magnitude of B:', manual_magnitude_b)
print('NumPy magnitude of B:', numpy_magnitude_b)

assert np.isclose(manual_magnitude_a, numpy_magnitude_a)
assert np.isclose(manual_magnitude_b, numpy_magnitude_b)
````

## 4. Cosine similarity

Cosine similarity measures the angle between two vectors:

$$\operatorname{cosine}(A,B)=\frac{A\cdot B}{\lVert A\rVert\lVert B\rVert}$$

Interpretation:

| Value | Meaning |
|---:|---|
| 1 | Same direction |
| Near 1 | Very similar direction |
| 0 | Unrelated or perpendicular |
| -1 | Opposite direction |

````python
manual_cosine = manual_dot_product / (manual_magnitude_a * manual_magnitude_b)
numpy_cosine = np.dot(vector_a, vector_b) / (
    np.linalg.norm(vector_a) * np.linalg.norm(vector_b)
)

print('Manual cosine similarity:', manual_cosine)
print('NumPy cosine similarity:', numpy_cosine)

assert np.isclose(manual_cosine, numpy_cosine)
assert np.isclose(numpy_cosine, 0.9746318461970762)
````

## 5. Create reusable cosine-similarity functions

````python
def manual_cosine_similarity(vector_1, vector_2):
    if len(vector_1) != len(vector_2):
        raise ValueError('Vectors must have the same length.')

    dot_product = sum(a * b for a, b in zip(vector_1, vector_2))
    magnitude_1 = math.sqrt(sum(value ** 2 for value in vector_1))
    magnitude_2 = math.sqrt(sum(value ** 2 for value in vector_2))

    if magnitude_1 == 0 or magnitude_2 == 0:
        raise ValueError('Cosine similarity is undefined for a zero vector.')

    return dot_product / (magnitude_1 * magnitude_2)


def numpy_cosine_similarity(vector_1, vector_2):
    vector_1 = np.asarray(vector_1, dtype=float)
    vector_2 = np.asarray(vector_2, dtype=float)

    if vector_1.shape != vector_2.shape:
        raise ValueError('Vectors must have the same shape.')

    denominator = np.linalg.norm(vector_1) * np.linalg.norm(vector_2)
    if denominator == 0:
        raise ValueError('Cosine similarity is undefined for a zero vector.')

    return float(np.dot(vector_1, vector_2) / denominator)


manual_result = manual_cosine_similarity([1, 2, 3], [4, 5, 6])
numpy_result = numpy_cosine_similarity([1, 2, 3], [4, 5, 6])

print('Manual function:', manual_result)
print('NumPy function:', numpy_result)
assert np.isclose(manual_result, numpy_result)
````

## 6. Cosine similarity with simple text vectors

Suppose the vocabulary is:

```text
[breaking, news, football, election]
```

Each number records how many times a word appears. This is a simplified bag-of-words representation.

````python
vocabulary = ['breaking', 'news', 'football', 'election']

text_vectors = {
    'breaking football news': np.array([1, 1, 1, 0], dtype=float),
    'football news update': np.array([0, 1, 1, 0], dtype=float),
    'election result': np.array([0, 0, 0, 1], dtype=float),
}

football_similarity = numpy_cosine_similarity(
    text_vectors['breaking football news'],
    text_vectors['football news update'],
)

election_similarity = numpy_cosine_similarity(
    text_vectors['breaking football news'],
    text_vectors['election result'],
)

print('Vocabulary:', vocabulary)
print('Related football texts:', football_similarity)
print('Football text vs election text:', election_similarity)

assert football_similarity > election_similarity
assert np.isclose(election_similarity, 0.0)
````

## 7. Pairwise similarity matrix

````python
labels = list(text_vectors.keys())
vector_matrix = np.vstack(list(text_vectors.values()))

norms = np.linalg.norm(vector_matrix, axis=1, keepdims=True)
normalized_matrix = vector_matrix / norms
similarity_matrix = normalized_matrix @ normalized_matrix.T

print('Labels:')
for index, label in enumerate(labels):
    print(index, label)

print('\nCosine-similarity matrix:')
print(np.round(similarity_matrix, 3))

assert similarity_matrix.shape == (3, 3)
assert np.allclose(np.diag(similarity_matrix), 1.0)
````
